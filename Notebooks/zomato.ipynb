# Zomato Bangalore Restaurants — Exploratory Data Analysis
This Jupyter Notebook loads, cleans, and explores your Bangalore restaurants CSV dataset (`address`, `name`, `online_order`, `book_table`, `rating`, `votes`, `location`, `cuisines`, `cost_per_person`, `type_of_restaurant`).

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='whitegrid')
plt.rcParams['figure.figsize'] = (10, 5)
```

## 1. Load the CSV File
Replace `'bangalore_restaurants.csv'` with the exact filename of your CSV file.

```python
FILE_PATH = 'bangalore_restaurants.csv'
df = pd.read_csv(FILE_PATH)

if 'Unnamed: 0' in df.columns:
    df = df.drop(columns=['Unnamed: 0'])

print(f'Dataset Shape: {df.shape[0]} rows, {df.shape[1]} columns')
df.head()
```

## 2. Data Cleaning & Summary Statistics

```python
df.info()

df['rating_clean'] = pd.to_numeric(df['rating'], errors='coerce').replace(0, np.nan)
df['cost_per_person'] = pd.to_numeric(df['cost_per_person'], errors='coerce')
df['votes'] = pd.to_numeric(df['votes'], errors='coerce')

print('\nMissing values per column:')
print(df.isnull().sum())
df.describe()
```

## 3. Top Locations & Restaurant Types

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

top_locations = df['location'].value_counts().head(10)
sns.barplot(x=top_locations.values, y=top_locations.index, ax=axes[0], palette='viridis')
axes[0].set_title('Top 10 Locations by Restaurant Listings')
axes[0].set_xlabel('Number of Listings')

type_counts = df['type_of_restaurant'].value_counts()
sns.barplot(x=type_counts.values, y=type_counts.index, ax=axes[1], palette='magma')
axes[1].set_title('Listings by Restaurant Type')
axes[1].set_xlabel('Number of Listings')

plt.tight_layout()
plt.show()
```

## 4. Rating & Cost Per Person Distributions

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.histplot(df['rating_clean'].dropna(), bins=20, kde=True, ax=axes[0], color='teal')
axes[0].set_title('Distribution of Ratings (Excluding Unrated)')
axes[0].set_xlabel('Rating')

sns.histplot(df['cost_per_person'].dropna(), bins=25, kde=True, ax=axes[1], color='coral')
axes[1].set_title('Distribution of Cost Per Person (INR)')
axes[1].set_xlabel('Cost Per Person (INR)')

plt.tight_layout()
plt.show()
```

## 5. Online Ordering, Table Booking & Top-Rated Restaurants

```python
print('Online Order Availability (%):')
print((df['online_order'].value_counts(normalize=True) * 100).round(2))

print('\nTable Booking Availability (%):')
print((df['book_table'].value_counts(normalize=True) * 100).round(2))

top_rated = (
    df[df['votes'] >= 500]
    .drop_duplicates(subset=['name', 'address'])
    .sort_values(by=['rating_clean', 'votes'], ascending=False)
    [['name', 'location', 'type_of_restaurant', 'rating_clean', 'votes', 'cost_per_person', 'cuisines']]
    .head(15)
)
top_rated
```
